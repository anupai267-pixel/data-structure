traversal

```python
arr = [10, 20, 30, 40, 50]

print("Array elements are:")

for i in range(len(arr)):
    print(arr[i])
```

insertion

```python
arr = [10, 20, 30, 40]
position = 2
element = 25
arr.insert(position, element)
print("Array after insertion:")
print(arr)
```

deletion

```python
arr = [10, 20, 30, 40, 50]
position = 2
arr.pop(position)
print("Array after deletion:")
print(arr)
```

linear search

```python
arr = [10, 20, 30, 40, 50]
key = 30
found = False
for i in range(len(arr)):
    if arr[i] == key:
        print("Element found at index", i)
        found = True
        break
if found == False:
    print("Element not found")
```

arrray update

```python
arr = [10, 20, 30, 40, 50]
position = 2
new_value = 35
arr[position] = new_value
print("Array after updation:")
print(arr)
```
