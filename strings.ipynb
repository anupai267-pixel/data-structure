Traverse a string

```python
string = "Python"
for i in range(len(string)):
    print(string[i])
```

length of string

```python
string = "Python"
length = len(string)
print("Length of the string is:", length)
```

palindrome

```python
string = input("Enter a string: ")
reverse = string[::-1]
if string == reverse:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

number of vowels

```python
string = input("Enter a string: ")
count = 0
for ch in string:
    if ch == 'a' or ch == 'e' or ch == 'i' or ch == 'o' or ch == 'u' or \
       ch == 'A' or ch == 'E' or ch == 'I' or ch == 'O' or ch == 'U':
        count = count + 1
print("Number of vowels:", count)
```
